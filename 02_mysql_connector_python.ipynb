## Connector
- db연결 정보를 작성

## Cursor
- SQL을 수행하고, 결과를 반환 받는 객체

**두 객체 모두 사용 후 반남(close) 처리 필수`**



```python
import mysql.connector

# DB연결정보 작성
config = {
    'host': 'localhost', #서버주소
    'port': 3306, #기본값이 3306,3306인 겨우 생략 가능
    'user': 'skn_ai',
    'password': '1234',
    'database': 'menudb',
}

conn = mysql.connector.connect(**config)

# 커서 객체 생성
cursor = conn.cursor()

# select 수행
# [주의] SQL 마지막에 세미클론(;) 작성 X
cursor.execute('select * from tbl_menu')
rows = cursor.fetchall() # 결과 집합(ResultSet) 반환

for row in rows :
    print(row)

# 사용 완료된 db 객체 close(생성 역순)
cursor.close()
conn.close()
```

```python
# 한 행을 표현하기 위한 Menu 클래스 정의
class Menu:
    def __init__(self,
                 menu_code, menu_name, menu_price, category_code, orderable_status):
        self.__menu_code = menu_code
        self.__menu_name = menu_name
        self.__menu_price = menu_price
        self.__category_code = category_code
        self.__orderable_status = orderable_status

    # -getter 역할을 하는 메서드
    #- 외부 호출 시 ()생략 -> menu.menu_code

    @property
    def menu_code(self):
        return self.__menu_code
    @property
    def menu_name(self):
        return self.__menu_name
    @property
    def menu_price(self):
        return self.__menu_price
    @property
    def category_code(self):
        return self.__category_code
    @property
    def orderable_status(self):
        return self.__orderable_status

    # __repr__() 메서드 : 객체에 어떤 값이 저장되어있는지 하나의 문자열로 반환
    # repr == representation(표현)
    def __repr__(self):
        return f"Menu(menu_code={self.__menu_code}, menu_name={self.__menu_name}, menu_price={self.__menu_price}, category_code={self.__category_code}, orderable_status={self.__orderable_status})"
```

```python
# 메뉴 테이블 전체 조회해서 Menu 리스트에 담기
try :
    with mysql.connector.connect(**config) as conn :
        with conn.cursor() as cursor :
            cursor.execute("select * from tbl_menu")
            rows = cursor.fetchall()  # 분리
            #조회 결과를 Menu 객체 List로 변환
            menus = [Menu(*row) for row in rows]

    for menu in menus:
        print(menu)
except mysql.connector.Error as err :
    print('DB 오류 : ',err)
```
